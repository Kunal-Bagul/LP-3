# Assignment 5: K-Means Clustering on Sales Data

### Step 1: Import required libraries

```python
# Import libraries for data handling, clustering, and visualization
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

### Step 2: Load sales dataset and select numerical columns

```python
# Read sales data with latin1 encoding and select numerical features
df = pd.read_csv("sales_data_sample.csv", encoding="latin1")
X = df[["SALES", "QUANTITYORDERED", "PRICEEACH"]].dropna()
X.head()
```

### Step 3: Normalize features using StandardScaler

```python
# Standardize data to zero mean and unit variance
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### Step 4: Use Elbow Method to determine optimal number of clusters

```python
# Compute WCSS (inertia) for k from 1 to 10 and plot elbow curve
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

### Step 5: Train final K-Means model with chosen k and assign clusters

```python
# Train K-Means with optimal k=3 and assign cluster labels
k = 3
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_scaled)

print(df[["SALES", "QUANTITYORDERED", "PRICEEACH", "Cluster"]].head())
print("\nCluster record counts:\n", df["Cluster"].value_counts())
```
