# Assignment 4: Gradient Descent to Find Local Minimum

### Step 1: Import required libraries

```python
# Import matplotlib for visualization
import matplotlib.pyplot as plt
```

### Step 2: Initialize starting point, learning rate, and iterations

```python
# Function: y = (x + 3)^2, Derivative: dy/dx = 2 * (x + 3)
x = 2
learning_rate = 0.1
iterations = 20
x_history = [x]
```

### Step 3: Run gradient descent loop to find local minimum

```python
# Update x using gradient descent: x = x - learning_rate * gradient
for i in range(iterations):
    gradient = 2 * (x + 3)
    x = x - learning_rate * gradient
    x_history.append(x)
    print(f"Iteration {i+1}: x = {x:.4f}, y = {(x + 3)**2:.4f}")

print(f"\nLocal minimum at x = {x:.4f}, y = {(x + 3)**2:.4f}")
```

### Step 4: Plot the optimization steps over iterations

```python
# Plot x values converging across iterations
plt.plot(x_history, marker="o")
plt.xlabel("Iteration")
plt.ylabel("x value")
plt.title("Gradient Descent Convergence towards x = -3")
plt.grid(True)
plt.show()
```
